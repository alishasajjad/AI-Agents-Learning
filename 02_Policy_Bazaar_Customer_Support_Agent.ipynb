# First AI Agent — Voice Customer Support Agent 🎙️

## Overview

The first AI agent built in this learning journey is a **Voice Customer Support Agent** for **Policybazaar**.

The purpose of this agent is to provide customers with information about insurance policies through a natural voice conversation.

The agent can understand customer questions, retrieve relevant information from the provided knowledge source, and respond through voice.

---

## Agent Goal

The primary goal of the agent is to:

* Answer insurance-related questions
* Provide information about different insurance policies
* Help customers understand available policies
* Assist customers in comparing and choosing suitable insurance options
* Reduce the need for human customer support
* Provide information based on the company's available knowledge

The agent should **not provide false promises or incorrect information**.

During the agent creation process, the following configuration was used:

```text
Agent Type: Business Agent
Business Category: Healthcare / Medical
Use Case: Customer Support
Agent Name: Policybazaar Customer Support Agent
Agent Type: Voice Agent
Language: English
```

---

## System Prompt

The agent was configured to behave as a professional AI voice agent for Policybazaar.

Its primary responsibility is to answer customers' questions related to:

* Insurance
* Health insurance
* Insurance policies
* Policy coverage
* Policy-related information

The agent should maintain a professional and helpful tone while providing accurate information.

It should avoid:

* Making false promises
* Providing incorrect information
* Making unsupported claims
* Giving information that is not available in the provided knowledge source

---

## Knowledge Base

A **Knowledge Base** was added to the agent so that it could answer questions using information from the Policybazaar website.

### Knowledge Source

**Policybazaar Website**

The website URL was added to the agent's knowledge base so that the agent could retrieve relevant information when answering customer questions.

```text
User Question
      ↓
Voice Agent
      ↓
Knowledge Base
      ↓
Retrieve Relevant Information
      ↓
LLM
      ↓
Generate Response
      ↓
Voice Response
```

---

## LLM

The agent configuration used:

```text
LLM: Gemini 2.5 Flash
Language: English
```

The LLM is responsible for understanding the customer's question and generating an appropriate response based on the available context and knowledge.

---

## How the Agent Works

The overall workflow of the voice customer support agent is:

```text
Customer speaks
       ↓
Speech is processed
       ↓
Agent understands the query
       ↓
Relevant information is retrieved
       ↓
LLM processes the information
       ↓
Agent generates a response
       ↓
Response is converted to voice
       ↓
Customer hears the answer
```

---

## Example Conversation

### Customer Query

> "I would like to know more about the health insurance policies available on Policybazaar."

### Agent

The agent can respond with information about available health insurance options and ask relevant follow-up questions, such as whether the customer is looking for coverage for themselves, their family, or parents.

### Another Query

> "I want to get a health insurance policy for my parents. Can you suggest some suitable options?"

The agent can ask for additional information, such as:

* Existing health conditions
* Desired coverage amount
* Customer requirements

This helps the agent provide more relevant information.

## Business Value

A voice customer support agent can reduce repetitive manual customer-support tasks.

Instead of requiring a human representative to answer every basic insurance-related question, the AI agent can handle common queries automatically.

```text
Traditional Support
Customer → Human Agent → Information

AI Voice Support
Customer → AI Agent → Knowledge Base → Response
```

This can help businesses:

* Automate repetitive queries
* Provide support at scale
* Reduce human workload
* Improve response speed
* Provide consistent information

## Key Learning

This project demonstrates how a **voice AI agent** can combine multiple AI-agent components:

```text
Goal
 ↓
LLM
 ↓
Knowledge Base
 ↓
Decision Making
 ↓
Voice Interaction
 ↓
Customer Support
```

The important concept learned from this project is that an AI agent is not limited to generating text. It can interact with users through voice, access external knowledge, understand requests, and provide automated assistance.
