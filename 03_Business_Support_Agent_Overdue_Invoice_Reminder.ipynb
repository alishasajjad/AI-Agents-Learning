# Business Support Agent — Overdue Invoice Payment Reminder

## Overview

In this module, we will build an **AI Business Support Agent** that automatically checks customer invoices, identifies overdue unpaid payments, sends reminder emails, and updates the invoice spreadsheet after the reminder has been sent.

The agent eliminates repetitive manual work and allows businesses to automate their **Accounts Receivable payment reminder workflow**.

---

## What This Agent Does

The agent performs the following workflow:

1. Runs automatically every day at a scheduled time.
2. Reads customer invoice data from Google Sheets.
3. Finds invoices that are:

   * Unpaid
   * Overdue
   * Not already reminded
4. Creates a polite payment reminder email.
5. Sends the email through Gmail.
6. Updates the Google Sheet to record that the reminder was sent.
7. Runs automatically without requiring continuous human involvement.

---

## Tools Used

The agent uses three main tools:

* **Google Sheets — Get Many Spreadsheet Rows (Advanced)**
  Used to retrieve invoice and customer data.

* **Gmail — Send Email**
  Used to send payment reminder emails to customers.

* **Google Sheets — Update Spreadsheet Row**
  Used to update the invoice record after the email has been sent.

## Step 1: Create the Agent

Open **Zapier Agents** and create a new agent.

You can either:

* Choose an existing template, or
* Create a **Custom Agent**

For this workflow, select **Create a Custom Agent** because our invoice reminder requirements are specific.

---

## Step 2: Define the Agent Instructions

The agent can be given instructions similar to:

```text
You are an Accounts Receivable Assistant for Alpha Company.

Every day after 2:00 PM, check the customer invoice spreadsheet.

Review all invoice records and identify customers whose payments are overdue.

For each overdue and unpaid invoice:

1. Send a polite payment reminder email.
2. Include the customer's first name.
3. Include the invoice number.
4. Include the outstanding amount.
5. Include the original due date.
6. Include the payment link.
7. After successfully sending the reminder, update the corresponding spreadsheet row to indicate that the reminder has been sent.

Do not send reminders for invoices that are not overdue or have already received a reminder.
```

The important part is to clearly define **what the agent should check, what action it should take, and what should happen afterward**.

---

## Step 3: Prepare the Google Sheet

The spreadsheet should contain the required customer and invoice information.

For example:

| Customer Name | Email                                               | Invoice Number | Product   | Amount | Due Date     | Payment Link | Status | Reminder Sent |
| ------------- | --------------------------------------------------- | -------------- | --------- | -----: | ------------ | ------------ | ------ | ------------- |
| Ali           | [customer@example.com](mailto:customer@example.com) | INV-001        | Product A | 25,000 | Aug 10, 2026 | Payment Link | Unpaid |               |
| Sara          | [sara@example.com](mailto:sara@example.com)         | INV-002        | Product B | 15,000 | Aug 20, 2026 | Payment Link | Unpaid |               |

The **Reminder Sent** column is important because it prevents the agent from repeatedly sending reminders for the same invoice.

---

## Step 4: Configure the Schedule

Set the agent to run automatically.

For example:

* **Frequency:** Every day
* **Time:** 2:00 PM
* **Weekend execution:** No
* **Time zone:** Asia/Kolkata

The schedule can be changed according to the business requirements.

Once configured, the workflow will automatically start at the specified time.

---

## Step 5: Add Google Sheets Tools

Go to **Add Tool** and add the Google Sheets tools.

### Tool 1 — Get Many Spreadsheet Rows

Select:

**Google Sheets → Get Many Spreadsheet Rows (Advanced)**

This allows the agent to retrieve multiple invoice records from the spreadsheet.

---

### Tool 2 — Update Spreadsheet Row

Add:

**Google Sheets → Update Spreadsheet Row**

This will allow the agent to update the invoice record after sending the reminder.

For example:

```text
Reminder Sent = 2026-08-16
```

---

## Step 6: Add Gmail

Next, add:

**Gmail → Send Email**

The agent will use Gmail to send payment reminder emails.

The agent can dynamically generate:

* Email recipient
* Subject
* Email body

based on the invoice information retrieved from Google Sheets.

---

## Step 7: Arrange the Tools in the Correct Order

The tools should follow the logical workflow:

```text
Google Sheets
     ↓
Find invoice records
     ↓
Check overdue + unpaid invoices
     ↓
Gmail
     ↓
Send payment reminder
     ↓
Google Sheets
     ↓
Update Reminder Sent
```

In other words:

### 1. Read the spreadsheet first

The agent needs invoice information before it can make a decision.

### 2. Send the reminder

If an invoice is overdue and unpaid, Gmail sends the reminder.

### 3. Update the spreadsheet

After the email is successfully sent, update the corresponding invoice row.

## Example Email

For an overdue customer, the agent can generate an email such as:

```text
Subject: Payment Reminder — Invoice INV-001

Dear Ali,

I hope this message finds you well.

This is a friendly reminder that payment for invoice INV-001
is currently overdue.

Amount Due: 25,000
Original Due Date: August 10, 2026

Please arrange payment at your earliest convenience using
the payment link below:

https://example.com/pay/1

If you have already made the payment, please disregard this
notice.

Thank you for your prompt attention to this matter.
```

The agent dynamically fills in the customer's information from the spreadsheet.

---

## Step 9: Verify the Email

After the agent sends the email, check Gmail to confirm that:

* The email was delivered.
* The customer's name is correct.
* The invoice number is correct.
* The amount is correct.
* The due date is correct.
* The payment link is included.
* The tone is professional and polite.

---

## Step 10: Verify the Spreadsheet Update

After the email is successfully sent, the agent updates the corresponding spreadsheet row.

For example:

**Before:**

```text
Reminder Sent: [Empty]
```

**After:**

```text
Reminder Sent: 2026-08-16
```

This allows the agent to know that the invoice has already received a reminder.

# Complete Workflow

The complete automation can be represented as:

```text
        Schedule Trigger
        Every Day at 2:00 PM
                 ↓
        Read Google Sheets
                 ↓
       Retrieve Invoice Rows
                 ↓
      Check Invoice Conditions
                 ↓
     ┌───────────┴───────────┐
     ↓                       ↓
Overdue + Unpaid          Not Overdue
     ↓                       ↓
Send Reminder              No Action
     ↓
Update Spreadsheet
     ↓
Reminder Sent = Date
     ↓
       Workflow Complete
```

## Human Involvement

This workflow can significantly reduce manual work.

Instead of an employee manually:

```text
Check spreadsheet
      ↓
Find overdue invoice
      ↓
Write email
      ↓
Send email
      ↓
Update spreadsheet
```

the AI agent can handle the process automatically:

```text
AI Agent
   ↓
Check invoices
   ↓
Identify overdue payments
   ↓
Send reminders
   ↓
Update records
```

Depending on the configured approval settings, some actions may still require human approval.
