# AI Business Support Agent with Langflow

## Overview

In this part, we build our **fourth AI Agent** using **Langflow**.

Langflow is a powerful **low-code platform** for building and deploying AI agents without requiring extensive coding. In this project, we create an **E-commerce Customer Support Agent** for a fictional company called **ShopEase**.

The agent can answer customer questions about products, refunds, warranties, shipping, returns, and other company-related information by using the company's knowledge base.

---

## 1. Agent Objective

The goal is to build an AI-powered **Business Support / Customer Support Agent** that can:

* Answer customer queries
* Provide product information
* Explain refund and return policies
* Provide warranty information
* Answer shipping-related questions
* Help customers track their orders
* Use company documents as a knowledge source
* Maintain conversation context using memory

### Example Queries

Customers can ask:

* "What is the refund policy?"
* "What products does ShopEase sell?"
* "What is the warranty period?"
* "Where is my order?"
* "Can I return my product?"

---

## 2. Company Knowledge Base

An AI model such as Gemini does not automatically know the private information of our company.

Therefore, we provide the agent with a **company-specific knowledge base**.

The knowledge base contains information such as:

* Product catalog
* FAQ
* Warranty policy
* Shipping policy
* Return policy
* Refund policy
* Other company documentation

These documents are provided as **PDF files**.

We also have an **order list** containing information such as:

* Customer name
* Order status
* Tracking ID
* Expected delivery date

This allows the agent to answer questions using our actual business information.

---

## 3. Install Langflow

First, open the Langflow website and create an account.

After signing up:

1. Download the Langflow application.
2. Select the appropriate operating system.
3. Install Langflow on your computer.
4. Open Langflow.
5. Sign in with your email.
6. Wait for the database and application setup to complete.

Once the installation is complete, the Langflow interface will be available.

---

## 4. Create a New Flow

After opening Langflow:

1. Click **New Flow**.
2. Select **Browse Mode**.
3. Choose a **Blank Flow**.

A blank workspace will be created where we can build our AI agent from scratch.

---

## 5. Add Chat Input

The first component we need is **Chat Input**.

Drag the **Chat Input** component into the workspace.

The Chat Input will receive messages from customers.

For example:

```text
What is ShopEase's refund policy?
```

The customer's message will then be passed to the AI Agent.

---

## 6. Add AI Agent

Next, add the **Agent** component.

Connect:

```text
Chat Input → Agent
```

The Agent will process the customer's question and generate an appropriate response.

---

## 7. Configure the Language Model

The AI Agent requires a Large Language Model (LLM).

Langflow provides multiple model providers, including:

* Anthropic
* Google
* IBM
* Ollama
* OpenAI
* OpenRouter

For this project, we use **Google Generative AI / Gemini**.

### API Key

To connect Gemini:

1. Open **Google AI Studio**.
2. Go to the API Keys section.
3. Create a new API key.
4. Copy the API key.
5. Return to Langflow.
6. Paste the API key into the Google Generative AI configuration.
7. Save the credentials.

The Gemini model can now be connected to the Agent.

---

## 8. Add Company Documents

Now we need to provide the AI Agent with the company's private knowledge.

From **Files & Knowledge**, add the **Read File** component.

Upload the company PDF documents containing information such as:

* Products
* Warranty
* Refund policy
* Return policy
* Shipping information
* FAQs

The agent can then access these documents when answering customer questions.

---

## 9. Connect Files as Tools

The uploaded files need to be available to the Agent.

Enable **Tool Mode** and connect the **Read File** component to the Agent's tools.

The basic flow becomes:

```text
Chat Input
     ↓
   Agent
     ↓
Gemini LLM
     ↓
Company Knowledge
```

The Agent can now retrieve relevant information from the company's documents.

---

## 10. Add Memory

To make the conversation more useful, we can add memory.

Use the **Message History** component.

Message History allows the agent to remember previous messages in the conversation and maintain context.

For example:

```text
Customer: What is your refund policy?

Agent: We offer a 30-day refund policy.

Customer: What about electronics?

Agent: The same policy applies to eligible electronics...
```

The second question can be understood in the context of the previous conversation.

Connect **Message History** with the Agent's tools and enable Tool Mode where required.

---

## 11. Test the Agent

After connecting all components, use the **Playground** to test the AI Agent.

Start a new session and send a simple message:

```text
Hello
```

The agent should respond with a customer-support greeting.

For example:

```text
Hello! Welcome to ShopEase Support.
How can I help you today?
```

---

## 12. Test Company Knowledge

Now test the agent using questions related to the uploaded documents.

### Refund Policy

Ask:

```text
What is ShopEase's refund policy?
```

The agent reads the relevant company document and provides the refund information.

### Products

Ask:

```text
What products does ShopEase sell?
```

The agent retrieves the relevant information from the product document.

For example, it may return products such as:

* Wireless keyboards
* Microphones
* Audio devices
* Fast chargers
* Power accessories
* Portable speakers

This demonstrates that the agent is using the **company knowledge base** rather than relying only on the model's general knowledge.

---

## 13. How the Agent Works

The overall architecture can be understood as:

```text
Customer
   ↓
Chat Input
   ↓
AI Agent
   ↓
Gemini LLM
   ↓
Company Knowledge Base
   ↓
PDF Documents / Message History
   ↓
Customer Response
```

The customer sends a query, the Agent processes it, Gemini understands the query, and the Agent retrieves relevant information from the company's knowledge base before generating the response.
