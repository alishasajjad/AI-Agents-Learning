# AI Startup Branding Agent with n8n

## Overview

In this project, we build a **no-code AI Startup Branding Agent** using **n8n** and **Google Gemini**.

The purpose of this agent is to help startup founders turn a basic business idea into a complete initial brand identity. Instead of manually brainstorming names, taglines, target audiences, positioning, and pitches, the AI agent generates these ideas automatically.

---

## 1. Agent Objective

Many people have a startup idea but struggle with questions such as:

* What should I name my startup?
* What tagline should I use?
* Who is my target audience?
* How should I position my brand?
* What should my logo look like?
* How can I explain my startup in a short pitch?

To solve this problem, we create an **AI Startup Branding Agent** that can handle multiple branding tasks from a single startup idea.

### Main Tasks

The agent can generate:

1. Startup name ideas
2. Taglines
3. Target audience
4. Brand positioning
5. Brand colors
6. Logo concepts
7. Domain name ideas
8. 30-second elevator pitches

---

# 2. Platform: n8n

The agent is built using **n8n**, a workflow automation platform that allows us to build AI-powered workflows with minimal or no traditional coding.

The general workflow is:

```text
User Message
     ↓
AI Agent
     ↓
Google Gemini Chat Model
     ↓
Branding Analysis
     ↓
Startup Branding Output
```

---

# 3. Creating the n8n Workspace

After signing up for n8n:

1. Complete the initial account setup.
2. Answer the workspace questions.
3. Choose the appropriate role and company information.
4. Select the option indicating that you already know what you want to automate.
5. Open the n8n workspace.
6. Create a new workflow.

n8n also provides pre-built templates for different automation use cases, but in this project we create a **custom AI agent**.

---

# 4. Creating the Workflow

Inside the n8n workspace:

1. Create a new workflow.
2. Add a trigger.
3. Add an **AI Agent** node.
4. Connect a chat model to the AI Agent.
5. Configure the AI Agent instructions.
6. Test the workflow using the chat interface.

The AI Agent becomes the main component responsible for understanding the user's startup idea and generating branding recommendations.

---

# 5. AI Agent Configuration

The AI Agent receives the user's message as input.

For example:

```text
I want to sell notes for IIT JEE exams.
This is my idea for my education startup.
Suggest me a startup name, tagline, and a 30-second pitch.
```

The agent analyzes this idea and generates relevant branding suggestions.

---

# 6. Connecting Google Gemini

For the AI model, we use **Google Gemini**.

### Steps

1. Add a Chat Model to the AI Agent.
2. Select **Google Gemini**.
3. Create or select the required credentials.
4. Provide the Gemini API key.
5. Save the credentials.
6. Connect the Gemini Chat Model to the AI Agent.

The architecture becomes:

```text
Chat Input
     ↓
AI Agent
     ↓
Google Gemini
     ↓
AI-generated response
```

### Getting the Gemini API Key

The API key can be generated through **Google AI Studio**.

General process:

```text
Google AI Studio
       ↓
API Keys
       ↓
Create API Key
       ↓
Copy API Key
       ↓
Add to n8n Credentials
```

> **Security Note:** Never share your Gemini API key publicly or upload it to GitHub.

---

# 7. AI Agent System Prompt

The agent can be configured with instructions such as:

```text
You are an Expert Startup Branding Consultant.

Your job is to create a complete brand identity based on the startup idea provided by the user.

Always use the user's latest message as the startup idea.

Do not ask the user to provide the startup idea again if they have already provided it.

If the user provides a clear startup idea, immediately generate the branding output.

For every valid startup idea, generate:

1. Startup Name – Provide 5 creative and professional options.
2. Domain Name Ideas – Provide 3-5 relevant domain name ideas.
3. One-line Tagline – Create one memorable tagline.
4. Target Audience – Clearly identify the primary customers.
5. Brand Positioning – Explain how the startup should be positioned.
6. Brand Colors – Suggest a professional color palette with reasons.
7. Logo Concept – Describe a simple and memorable logo concept.
8. 30-second Elevator Pitch – Write a natural and convincing pitch.

Do not repeat:
"Please provide your startup idea"
"Please share your startup idea"
"I am ready to help"

when the user has already provided a valid startup idea.

Return the final branding output directly.
```

# 8. Testing the Agent

After connecting the Gemini model and configuring the prompt, use the **Open Chat** option to test the agent.

First, a simple message can be tested:

```text
Hi
```

The agent should respond appropriately.

Then provide an actual startup idea.

### Example Input

```text
I want to sell notes for IIT JEE exams.

This is my idea for my education startup.

Suggest me:
- A startup name
- One tagline
- A 30-second pitch
```

The message is sent to the AI Agent, which passes the request to the Google Gemini model.

# 10. How the Agent Works

The complete process is:

```text
User provides startup idea
          ↓
n8n Chat Trigger
          ↓
AI Agent
          ↓
Google Gemini Chat Model
          ↓
Understands the startup idea
          ↓
Generates branding recommendations
          ↓
Returns final response
```

The important part is that the AI Agent doesn't perform only one task.

It can perform **multiple branding tasks from a single user request**.
